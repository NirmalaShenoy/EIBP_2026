## Access the Slice

The orchestrator class is initalized, which also means the slice and its nodes are now accessable as well.

```python
from FabUtils import FabOrchestrator

try:
    manager = FabOrchestrator(SLICE_NAME)

except Exception as e:
    print(f"Exception: {e}")
```

```python
manager.executeCommandsParallel("ip address | grep '<BROADCAST,MULTICAST>'")
```

```python
for node_name in manager.slice.nodes:
    node = manager.slice.get_node(node_name)
    print(node.get_name())
```

```python
manager.executeCommandsParallel("cat .tmux.conf && cat tmux_start_logging.sh")
```

```python
manager.executeCommandsParallel("tmux ls")
```

```python
# Grab the IP Routing table from each compute node.
print("ROUTING TABLES ON COMPUTE NODES\n")

routingTableCommand = "ip route"
manager.executeCommandsParallel(routingTableCommand, prefixList="ipnode")
```

```python
# Grab the interface status from each node (regardless of the type of node).
print("NETWORK INTERFACE STATUS\n")

intfStatusCmd = "ip address | grep eth"
manager.executeCommandsParallel(intfStatusCmd, prefixList="C,D,A")
```

```python
# Find any interfaces that are set to down
print("CHECK IF INTERFACES ARE DOWN\n")

intfStatusCmd = "ip address | grep DOWN"
manager.executeCommandsParallel(intfStatusCmd, prefixList="C,D,A")
```
