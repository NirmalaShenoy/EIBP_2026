## Take the Interface down

This code brings down a network interface specified by {INTF_TO_FAIL} across multiple nodes and then retrieves IP addresses from the specified nodes in parallel using Python's manager.executeCommandsParallel function.

```python
def apply_soft_failure(node, intf):

    # Remove existing table if present
    cleanup_cmd = "sudo nft delete table netdev starvation 2>/dev/null || true"
    manager.executeCommandsParallel(cleanup_cmd, prefixList=node)

    # Create table
    create_table_cmd = "sudo nft add table netdev starvation"
    manager.executeCommandsParallel(create_table_cmd, prefixList=node)

    # Create ingress chain on the interface
    create_chain_cmd = (
        f'sudo nft add chain netdev starvation {intf}_in '
        f'"{{{{ type filter hook ingress device {intf} priority 0; policy accept; }}}}"'
    )
    manager.executeCommandsParallel(create_chain_cmd, prefixList=node)

   
    drop_with_time_cmd = (
        f"sudo nft add rule netdev starvation {intf}_in drop && "
        f"date +%s.%6N > Soft_Intf_Down.txt"
    )
    manager.executeCommandsParallel(drop_with_time_cmd, prefixList=node)
```
